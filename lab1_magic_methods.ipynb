## Introduction

Python's **magic methods** (also called dunder methods) let you define what operators do when applied to your custom classes. This is the foundation of micrograd: a `Value` object behaves just like a number, but secretly records every operation it's involved in.

### Key Concepts

- `__add__(self, other)` → called when you write `self + other`
- `__mul__(self, other)` → called when you write `self * other`
- `__repr__(self)` → called when you `print` the object

**KEY INSIGHT**: Even built-in numbers are objects — `int.__add__` is the magic method that defines what `3 + 5` means. micrograd gives us smarter numbers that remember their history.

```python
class Matrix:
    def __init__(self, data):
        self.data = data
        self.rows = len(data)
        self.cols = len(data[0])

    def __add__(self, other):
        assert self.rows == other.rows and self.cols == other.cols
        result = [
            [self.data[i][j] + other.data[i][j] for j in range(self.cols)]
            for i in range(self.rows)
        ]
        return Matrix(result)

    def __mul__(self, other):
        assert self.cols == other.rows
        result = [
            [
                sum(self.data[i][k] * other.data[k][j] for k in range(self.cols))
                for j in range(other.cols)
            ]
            for i in range(self.rows)
        ]
        return Matrix(result)

    def __repr__(self):
        return '\n'.join(str(row) for row in self.data)
```
